## Reflections on Quantum Machine Learning — Honest Observations and a Path Forward

### The "Quantum-Inspired" Reality

After working extensively with quantum circuits, I have come to believe that a significant portion of what is currently called "quantum machine learning" would be more accurately described as **quantum-inspired machine learning**. My reasoning is straightforward: if a parameterized quantum circuit can be simulated efficiently on a classical CPU or GPU, and we observe benefits such as faster convergence or fewer parameters compared to a conventional neural network, this does not necessarily demonstrate a quantum advantage. It more likely indicates that we have found a different — and perhaps better — mathematical parametrization that classical hardware can handle perfectly well.

Many published works in this space report results obtained entirely from classical simulators, without comparing against execution on actual quantum hardware. Without that comparison, claims about computational speedup remain unverifiable. When we decompose most parameterized quantum circuits into their underlying matrix operations, what we find is standard linear algebra — rotations, tensor products, and unitary transformations — all of which classical computers are very efficient at performing.

That said, I do not dismiss the field entirely. I believe there are genuine cases where quantum computing can offer something classical approaches cannot. The key is being honest about where that boundary lies.

### Software Ecosystem: From Qiskit to CUDA-Q

My personal journey through quantum software has followed the path: **Qiskit → PennyLane → CUDA-Q**. This evolution reflects my growing conviction that the future of quantum computing lies in **HPC–QC integration** — tightly coupling classical high-performance computing with quantum processors, rather than treating quantum devices as standalone systems.

CUDA-Q, developed by NVIDIA, aligns well with this vision. It has established partnerships with multiple quantum hardware vendors. For instance, during my time as a visiting researcher in Prof. Alán Aspuru-Guzik's laboratory (The Matter Lab), we used Quantinuum's quantum hardware through such integrations. In contrast, Qiskit has become increasingly closed in recent years, with fewer collaborations with external communities and hardware providers. For researchers who want flexibility across different quantum backends, an open and hardware-agnostic ecosystem matters.

### What I Believe Truly Useful QML Looks Like

Based on my research experience, I believe the most productive direction for quantum machine learning is not to replace classical models with quantum ones, but rather to build **hybrid systems where each paradigm strengthens the other**:

- **Using classical ML to assist quantum computing**: During my undergraduate studies, I used LSTMs to predict good initial parameters for the Variational Quantum Eigensolver (VQE), reducing the optimization burden on the quantum side. More recently, at The Matter Lab, I have learned how large language models can be trained to **design quantum circuits directly** — rather than tuning rotation angles within a fixed circuit ansatz, we reframe the optimization as training a Transformer to generate circuit architectures. This is a fundamentally different approach: the neural network learns the structure of the solution, not just its parameters.

- **Using quantum components to enhance classical models**: Our current research applies this circuit-generation technique to augment the expressiveness of classical generative models such as VAEs and LSTMs for molecular design — specifically targeting cancer drug candidates and novel materials. I am currently responsible for generating **Quantum Circuit Born Machine (QCBM)** circuits within this pipeline.

The power of this approach is its generality: the same Transformer-based circuit generation framework can be directed toward quantum chemistry circuits, quantum machine learning tasks, or any domain where the circuit structure itself is the object of optimization.

### Closing Thoughts

I believe the quantum machine learning community would benefit from more intellectual honesty about what constitutes genuine quantum advantage versus what is simply a useful alternative mathematical framework. At the same time, the intersection of classical AI and quantum computing — where each side compensates for the other's weaknesses — holds real promise. This is the space I am most excited to contribute to, and it aligns closely with the goals of the QML-HEP projects.
